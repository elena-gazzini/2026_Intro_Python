## Exercise 3: Operator families and a quality flag

Three daily readings arrive from a station: `reading_1_celsius` = 18.2, `reading_2_celsius` = -300.0, and `reading_3_celsius` = None (the sensor reported nothing).

1. Build a boolean `is_plausible` that is `True` when the first reading is above absolute zero and below 60 °C.
2. Build `is_implausible` for the second reading: apply the same test as in step 1, and negate it with `not`.
3. Use `is` to check whether the third reading is missing.
4. Starting from `n_valid = 0`, use `+=` twice to count how many of the first two readings pass the step 1 test. Note that `int()` casts `True` to 1 and `False` to 0.
5. The sensor is rated for -40 °C to 60 °C, both bounds included. Use `>=` and `<=` to test whether the first reading lies in that range, and `!=` to check that the first two readings differ.

````python
# naming the values
reading_1_celsius = 18.2
reading_2_celsius = -300.0
reading_3_celsius = None

# 1. Build a boolean is_plausible that is True when the first reading is above absolute zero and below 60 °C.
is_plausible = reading_1_celsius > 0 and reading_1_celsius < 60
print(f"is_plausible: {is_plausible}")

# 2. Build is_implausible for the second reading: apply the same test as in step 1, and negate it with not.
is_implausible = not(reading_1_celsius > 0 and reading_1_celsius < 60)
# is_implausible = not reading_1_celsius > 0 and not reading_1_celsius < 60
# both same results
print(f"is_implausible: {is_implausible}")

# 3. Use is to check whether the third reading is missing.
missing = reading_3_celsius is None
print(f"missing: {missing}")

# 4. Starting from n_valid = 0, use += twice to count how many of the first two readings pass the step 1 test. Note that int() casts True to 1 and False to 0.
n_valid = 0

n_valid += int(is_plausible)
# same as n_valid += int(reading_1_celsius > 0 and reading_1_celsius < 60)
n_valid += int(reading_2_celsius > 0 and reading_2_celsius < 60)

print(f"n_valid: {n_valid}")

# 5. The sensor is rated for -40 °C to 60 °C, both bounds included. Use >= and <= to test whether the first reading lies in that range, and != to check that the first two readings differ.
sensor_range = (reading_1_celsius >= -40 and reading_1_celsius <= 60)
print(f"sensor_range: {sensor_range}")

sensor_different = reading_1_celsius != reading_2_celsius
print(f"sensor_different: {sensor_different}")
````

## Exercise 4: Parse a record

A logger emits records like `"2024-01-15 , JUNGFRAUJOCH , -2.3 , degC"`. Extract the date string, the station name in title case, the numeric value as a float, and the unit as written, then print them. Then print the station name in lower case as well, the form you would use in a file name.

````python

record = "2024-01-15 , JUNGFRAUJOCH , -2.3 , degC"

# Extract each field (first we have a date, then a station, a numeric value and a unit)
date_string, station_name, numeric_value, unit = record.split(",") # slipt the record when there is a comma

# Clean the values : .strip() removes unnecessary spaces around the text.
date_string = date_string.strip()
station_name = station_name.strip().title() # .title() transforms JUNGFRAUJOCH to Jungfraujoch.
numeric_value = float(numeric_value.strip()) # float() transforms "-2.3" in a number -2.3
unit = unit.strip()

# Print the extracted values
print(f"Date: {date_string}")
print(f"Station: {station_name}")
print(f"Value: {numeric_value}")
print(f"Unit: {unit}")


# Station name in lowercase for a file name
print(f"Station filename: {station_name.lower()}") # .lower() transforms Jungfraujoch to jungfraujoch
````

## Exercise 5: Indexing and slicing a record

Weather records often arrive in a fixed shape, where each field always occupies the same
positions. Given

```python
record = "ALO01.03.2022"
```

the first three characters are the station code and the rest is a date as `dd.mm.yyyy`.

1. Using **only indexing and slicing** (no `.split()`), extract the station code, the day, the

2. Now extract the month a second way, using `.split(".")` on the date part, and use `==` to

   record where the station code can be two or four characters long, and why.

````python
record = "ALO01.03.2022"

# 1. Using only indexing and slicing
station_code = record[:3] # take only the first 3 letters ; python starts from 0
print(f"station_code: {station_code}")

day = record[3:5] # take only what's in the places 3 and 4 (not 5)
print(f"day: {day}")

month = record[6:8]
print(f"month: {month}")

year = record[9:]
# same as year = record[9:13]
print(f"year: {year}")

# 2. Extract the month using split()
date_part = record[3:]
print(f"date_part: {date_part}")

date_parts = date_part.split(".")
print(f"date_parts: {date_parts}")

month_from_split = date_parts[1] # 1 because begins at 0

print(f"Month from split: {month_from_split}")
print(f"Same month: {month == month_from_split}")

day_from_split = date_parts[0]
print(f"Day from split: {day_from_split}")
print(f"Same day: {month == day_from_split}")

# 3. I would trust split(".") if the station code can be two or four characters long,
# because fixed-position slicing depends on the station code always having the same length.
````

### Exercise 8: Same seven values, same mean?

Adding the same floats in a different order does not have to give exactly the same result,
because each `+` rounds its result to the nearest float. This exercise uses `math.isclose` from
the going-deeper box on floating-point precision.

1. Compute the mean of the Jungfraujoch week — -8.4, -8.9, -9.2, -9.0, -9.5, -7.8, -6.1 — by
   adding the values in that order and dividing by 7.
2. Compute it again, adding the same seven values in reverse order.
3. Compare the two means with `==`, and print the result.
4. Compare the same two means with `math.isclose`, and print the result.
5. In a comment, say which of the two comparisons you would put in a script that checks whether
   two calculations of the same quantity agree, and why.

````python
# 1.additionner les 7 températures dans l'ordre donné :
mean_forward = (-8.4 + -8.9 + -9.2 + -9.0 + -9.5 + -7.8 + -6.1) / 7
print(mean_forward) # mean_forward contient la moyenne obtenue dans cet ordre

# 2. idem mais à l'envers :
mean_reverse = (-6.1 + -7.8 + -9.5 + -9.0 + -9.2 + -8.9 + -8.4) / 7
print(mean_reverse)
````

Mathématiquement, c'est exactement les mêmes nombres, donc les deux résultats doivent être identiques.

Et en théorie mathématique, oui.

Mais Python travaille avec des nombres flottants (`float`), qui ne représentent pas parfaitement tous les nombres décimaux. À chaque addition, il peut donc y avoir une minuscule différence d'arrondi.

````python
# 3. Avec ==
print(mean_forward == mean_reverse)
````

`==` demande « Est-ce que les deux nombres sont exactement identiques ? »

Il est possible que Python réponde : `False`, même si la différence est minuscule.

````python
# 4. Avec math.isclose()
import math

print(math.isclose(mean_forward, mean_reverse))
````

`math.isclose()` pose plutôt la question : « Est-ce que les deux nombres sont suffisamment proches pour qu'on puisse considérer qu'ils représentent le même résultat ? »

Dans ce cas, il devrait répondre : `True`.

Dans un script scientifique qui vérifie si deux calculs donnent le même résultat, tu utiliserais plutôt : `math.isclose()`.

Parce que tu ne veux généralement pas que ton programme considère deux résultats comme différents simplement à cause d'une infime différence d'arrondi des `float`.

Donc l'idée à retenir est :

`==`             → exactement pareil ?

`math.isclose()`  → suffisamment proche ?

Donc, la fonction `math.isclose` est plus adaptée pour comparer des valeurs à virgule flottante calculées, car de très légères différences d'arrondi peuvent survenir.

- `f.write("station,day,max_temp_celsius\n")` → écrit le titre des colonnes.
- `f.write(f"{station_code_1},{day_1},{max_temp_celsius_1:.1f}\n")` → écrit les informations du **premier jour**.
- `f.write(f"{station_code_2},{day_2},{max_temp_celsius_2:.1f}\n")` → écrit les informations du deuxième jour.

*rappel* Le `.1f` signifie une décimale.

3. Relire le fichier

Ensuite, rouvrir le fichier en mode `"r"` :

````python
with summary_path.open("r") as f:
    content = f.read()

print(content)
````
